## One paragraph per notebook — what you can say in an interview

- **hugging_face_chromadb_demo** — I built a pipeline from PDFs to chunked text to embeddings and a vector DB; that’s the data-engineering backbone of RAG.  
- **gemini_rag_pipeline_demo** — I retrieve sources, generate an answer, and run a second model as a judge to check grounding — production-style guardrail.  
- **ai_safety_evals_demo** — I compared Mahalanobis, centroid, and isolation forest for adversarial prompt detection with proper ROC and bootstrap CIs.  
- **ai_security_benchmarking_demo** — I implemented defense-in-depth and red-team benchmarks with Wilson intervals — how I’d document safety testing.  
- **rag_hallucination_scoring** — I combined retrieval, citation, and judge signals into a three-tier route (reliable / uncertain / unreliable).  
- **attention_hallucination_demo** — I used attention entropy and cross-layer KL as a fast, training-free hallucination signal on a small LM.  
- **self_data_hallucination_classifier** — I used LLM-as-judge labels to train a lightweight classifier — self-data scaling.  
- **tabular_claims_fraud_ml** — I worked imbalanced fraud labels with PR-AUC and balanced logistic regression — classic P&C ML.  
- **synthetic_policy_rag_walkthrough** — I chunked a fake policy and ran retrieval — same pattern as a coverage Q&A bot without real data.  
- **rag_eval_logging** — I append structured JSONL per query for observability and governance narrative.  
- **fairness_basics_demo** — I measured demographic parity differences on synthetic groups — I know fairness must be measured and governed.  
- **adversarial_prompt_crafting_lab** — Hands-on red-team intuition for injection and system-prompt leakage.
